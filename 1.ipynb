```python
# 3.1 Створіть функцію, яка на основі втрат і чисельності ворога оцінює результативність дій.
def evaluate_losses(a):
    if 0 <= a <= 20:
        return "Мінімальні втрати"
    elif 21 <= a <= 40:
        return "Невеликі втрати"
    elif 41 <= a <= 60:
        return "Середні втрати"
    elif 61 <= a <= 80:
        return "Великі втрати"
    elif 81 <= a <= 100:
        return "Катастрофічні втрати"
    else:
        return "Некоректне значення"


```

```python
# 3.2. Напишіть функцію для обрахунку середніх втрат за кілька днів.

def average_losses(day1, day2, day3, day4, day5):
    average = (day1 + day2 + day3 + day4 + day5) / 5
    return average
```

```python
day1 = int(input("Введіть втрати за 1 день: "))
day2 = int(input("Введіть втрати за 2 день: "))
day3 = int(input("Введіть втрати за 3 день: "))
day4 = int(input("Введіть втрати за 4 день: "))
day5 = int(input("Введіть втрати за 5 день: "))

result = average_losses(day1, day2, day3, day4, day5)

print("Середні втрати за 5 днів:", result)
```

```python
def assess_fuel(percent):
    if percent >= 70:
        return "Готовий"
    elif percent >= 50:
        return "Обмежено готовий"
    else:
        return "Не готовий"


def assess_ammo(percent):
    if percent >= 70:
        return "Готовий"
    elif percent >= 50:
        return "Обмежено готовий"
    else:
        return "Не готовий"


def assess_food(percent):
    if percent >= 70:
        return "Готовий"
    elif percent >= 50:
        return "Обмежено готовий"
    else:
        return "Не готовий"


def average_readiness(fuel, ammo, food):
    return (fuel + ammo + food) / 3
```

```python
fuel = float(input("Введіть рівень забезпечення пальним (%): "))
ammo = float(input("Введіть рівень забезпечення боєприпасами (%): "))
food = float(input("Введіть рівень забезпечення продовольством (%): "))

print("Пальне:", assess_fuel(fuel))
print("Боєприпаси:", assess_ammo(ammo))
print("Продовольство:", assess_food(food))

result = average_readiness(fuel, ammo, food)

print("Середній рівень забезпечення:", result, "%")

if result >= 70:
    print("Загальна оцінка: Готовий")
elif result >= 50:
    print("Загальна оцінка: Обмежено готовий")
else:
    print("Загальна оцінка: Не готовий")
```
