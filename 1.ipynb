```python
print("Задача 1.Вивести повідомлення про рівень втрат противника")
print("        Якщо число від 0 до 20 – “Мінімальні втрати”") 
print("        Якщо число від 21 до 40 – “Невеликі втрати”") 
print("        Якщо число від 41 до 60 – “Середні втрати”")
print("        Якщо число від 61 до 80 – “Великі втрати”") 
print("        Якщо число від 81 до 100 – “Катастрофічні втрати”")  
print("") 
a = 12
print("Втрати противника складають", a, "%")
if a >= 0 and a <= 20:
    print("Мінімальні втрати")

elif a >= 21 and a <= 40:
    print("Невеликі втрати")
elif a >= 41 and a <= 60:
    print("Середні втрати") 
elif a >= 61 and a <= 80:
    print("Великі втрати")
elif a >= 81 and a <= 100:
    print("Катастрофічні втрати")
```
